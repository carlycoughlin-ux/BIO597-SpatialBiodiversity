```python
path_to_data = geodatasets.get_path("ny.bb")
gdf = gpd.read_file(path_to_data)
# putting a variable on the last line of a cell will ask jupyter to print a human readable version
# of the contents, if possible
gdf
```

```python
df = gpd.read_file("Sdekayi_coords.csv")

gdf = gpd.GeoDataFrame(
    df,
    geometry=gpd.points_from_xy(df["Longitude"], df["Latitude"]),
    crs="EPSG:4326"
)
gdf.explore()
```
